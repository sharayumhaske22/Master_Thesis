```python
import csv
from google.colab import files

# Step 1: Define file path (in Colab)
FILE = "/content/sample_file.csv"

# Step 2: Create file and write data
with open(FILE, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    
    # Write header
    writer.writerow(["Name", "Age", "City"])
    
    # Write some rows
    writer.writerow(["Alice", 25, "Berlin"])
    writer.writerow(["Bob", 30, "Paris"])
    writer.writerow(["Charlie", 22, "London"])

print("File created and written successfully")

# Step 3: Download file to your computer
files.download(FILE)
```
